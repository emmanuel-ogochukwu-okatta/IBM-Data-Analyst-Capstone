```python
import pandas as pd
d = pd.read_csv("survey_data_updated.csv")

cols = ["LanguageHaveWorkedWith", "DatabaseHaveWorkedWith", "PlatformHaveWorkedWith",
        "WebframeHaveWorkedWith", "LanguageWantToWorkWith", "DatabaseWantToWorkWith",
        "PlatformWantToWorkWith", "WebframeWantToWorkWith"]

for c in cols:
    real = next(x for x in d.columns if x.lower() == c.lower())
    out = (d[["ResponseId", real]].dropna()
             .assign(**{real: lambda t: t[real].str.split(";")})
             .explode(real))
    out[real] = out[real].str.strip()
    out.to_csv(f"split_{c}.csv", index=False)
    print(c, len(out), "rows")
```
